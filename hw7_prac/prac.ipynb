## Topic 1: Bit Manipulation Basics
Concept:

Understand binary representation and basic operations (AND, OR, XOR, NOT, shifts).

```python
# Q1.1: Check if a bit is set

def is_bit_set(num, pos):
    return (num & (1<<pos))!=0

is_bit_set(13,2)
```

```python
# Q1.2: Setting a bit to 1

def set_bit(num, pos):
    return (num | (1<<pos))

set_bit(9,1)
```

```python
#Q1.2: CLearing a bit at a position

def clear_bit(num, pos):
    return (num & ~(1<<pos))

clear_bit(15,3)
```
