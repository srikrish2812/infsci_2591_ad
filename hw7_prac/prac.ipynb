## Topic 1: Bit Manipulation Basics
Concept:

Understand binary representation and basic operations (AND, OR, XOR, NOT, shifts).

```python
# Q1.1: Check if a bit is set
# Intuition:
# 1 << position creates a mask with only that bit set
# AND checks if that bit exists in our number


def is_bit_set(num, pos):
    return (num & (1<<pos))!=0

is_bit_set(13,2)
```

```python
# Q1.2: Setting a bit to 1
# Intuition: OR forces a bit to be 1, regardless of its current value.
def set_bit(num, pos):
    return (num | (1<<pos))

set_bit(9,1)
```

```python
#Q1.3: CLearing a bit at a position
# Intuition:
# Create a mask with 0 at the position we want to clear
# AND forces that bit to 0

def clear_bit(num, pos):
    return (num & ~(1<<pos))

clear_bit(15,3)
```

```python
#Q1.4: Toggle a bit at a specific position
def toggle_bit(num: int, position: int) -> int:
    """
    Toggle (flip) the bit at 'position'.
    
    Example: toggle_bit(5, 0) → 4  (5 = 0b101, result = 0b100)
             toggle_bit(4, 0) → 5
    """
    return num ^ (1<<position)

toggle_bit(5,0)
```

## Topic 2: Essential Bit Tricks

```python
#Q2.1: Isolate the rightmost set bit
#Intuition:
# -n in two's complement creates a mask where only the rightmost 1 and everything to its right are preserved
# AND extracts just that bit

def isolate_rightmost_bit(num):
    return num & (-num)

isolate_rightmost_bit(13)
```

```python
#Q2.2: Remove the rightmost set bit
# Intuition: Subtracting 1 flips all bits from the rightmost 1 to the end. AND removes that bit.
# Use case: Counting set bits efficiently
def remove_rightmost_setbit(num):
    return num & (num-1)

remove_rightmost_setbit(13)
```

```python
#Q2.3: Count the number of set bits
def count_set_bits(num: int) -> int:
    """
    Count the number of 1s in the binary representation.
    
    Example: count_set_bits(13) → 3  (13 = 0b1101)
    """
    count =0
    while num:
        num &= (num-1)
        count+=1
    return count

count_set_bits(13)
```

```python
#Q2.4: Check if a nyumber is a power fo two
def is_power_of_two(num: int) -> bool:
    """
    Return True if num is a power of 2.
    
    Example: is_power_of_two(8) → True
             is_power_of_two(6) → False
    """
    return num>0 and (num & (num-1))==0 and num!=1


is_power_of_two(6)
```
